Pydantic but More Complex

```python
from pydantic import BaseModel
from typing import List, Dict, Optional
```

```python
class PatientData(BaseModel):
    name: str
    age: int
    weight : float
    married : bool = False
    allergies : Optional[List[str]] = None
    contacts_info : Dict[str, str]
    
def add_patient_data(patient:PatientData):
    print(f"Patient Name: {patient.name}, Age: {patient.age}, Weight: {patient.weight}, Married: {patient.married}, Allergies: {patient.allergies}, Contacts Info: {patient.contacts_info}")
    
def update_patient_data(patient:PatientData):
    print(f"Updated Patient Name: {patient.name}, Age: {patient.age}, Weight: {patient.weight}, Married: {patient.married}, Allergies: {patient.allergies}, Contacts Info: {patient.contacts_info}")
    
patient_data = {"name": "John Doe", "age": 35, "weight": 70.5, "married": True, "allergies": ["pollen", "dust"], "contacts_info": {"email": "john.doe@example.com", "phone": "123-456-7890"}}

patient_1 = PatientData(**patient_data)

add_patient_data(patient_1)
```

```python
patient_data =  {"name": "John Dog", "age": -36, "weight": 75.0, "married": False, "allergies": ["peanuts"], "contacts_info": {"email": "ArithmeticError@domain.com"}}
patient_2 = PatientData(**patient_data)

update_patient_data(patient_2)
```
