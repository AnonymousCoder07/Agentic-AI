```python
from pydantic import BaseModel

class PatientData(BaseModel):
    name: str
    age: int
    weight : float
    
def add_patient_data(patient: PatientData):
    print(f"Patient Name: {patient.name}, Age: {patient.age}, Weight: {patient.weight}")
    
def update_patient_data(patient: PatientData):
    print(f"Updated Patient Name: {patient.name}, Age: {patient.age}, Weight: {patient.weight}")

patient_data = {"name": "John Doe", "age": 35, "weight": 70.5}

patient_1 = PatientData(**patient_data)

add_patient_data(patient_1)
```

```python
patient_data = {"name": "John Dog", "age": -36, "weight": 75.0}
patient_2 = PatientData(**patient_data)

update_patient_data(patient_2)
```
